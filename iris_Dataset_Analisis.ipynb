Exploratory Data Analysis (EDA) of the Iris Dataset

Project Goal

The goal of this project is to explore and understand the structure of the Iris dataset. This includes loading the dataset, examining its features, checking for missing values, and visualizing the data to understand the distribution and relationships between variables.

Description

In this step, we import the necessary Python libraries used for data manipulation, analysis, and visualization.

Pandas is used for handling datasets.

NumPy helps with numerical operations.

Matplotlib and Seaborn are used for data visualization.

Scikit-learn provides the Iris dataset.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# For dataset
from sklearn.datasets import load_iris
```

```python
# Load the dataset
iris = load_iris()

# Convert to DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add target column
df['species'] = iris.target

# Show first rows
df.head()
```

In this step, we load the Iris dataset from the scikit-learn library and convert it into a pandas DataFrame. This allows us to easily explore and manipulate the dataset.

```python
# Dataset shape
print("Shape of dataset:", df.shape)

# Column names
print("Columns:", df.columns)

# Data types
df.dtypes
```

```python
# General information
df.info()
```

eHere we xamine the basic structure of the dataset, including the number of rows and columns, the column names, and the type of data stored in each column.

This step provides a statistical overview of the dataset, including the mean, minimum, maximum, and standard deviation of each feature. This helps us understand the general distribution of the data.

```python
# Statistical summary
df.describe()
```

Duplicate rows can sometimes exist in datasets and may affect analysis. In this step, we check if any duplicate records are present.

```python
# Check missing values
df.isnull().sum()
```

Before performing analysis, it is important to ensure that the dataset does not contain missing values. Missing data can affect the accuracy of analysis and machine learning models.

The output shows that there are no missing values in the dataset, meaning the data is already clean and ready for analysis.

df.duplicated().sum()

```python
df = df.drop_duplicates()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.boxplot(data=df)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

Data Cleaning

The dataset was checked for missing values and duplicate records. No missing values or duplicates were found in the dataset. Therefore, no rows were removed during the data cleaning process.

```python
# Check missing values
df.isnull().sum()

# Check duplicates
df.duplicated().sum()

# Remove duplicates if any
df = df.drop_duplicates()
```

Histograms help visualize the distribution of each feature in the dataset. This allows us to see how values are spread across the dataset.

```python
df.hist(figsize=(10,6))
plt.show()
```

A pairplot shows the relationships between different features of the dataset. It also helps visualize how different species of flowers are distributed based on these features.

```python
sns.pairplot(df, hue='species')
plt.show()
```

A correlation heatmap helps identify relationships between features. It shows how strongly different variables are related to each other.

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()
```

Boxplots help detect outliers and show the spread of the data for each feature

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.title("Feature Distribution with Boxplot")
plt.show()
```
